# Tourism Data Analysis & Travel Trend Visualization

```python
import pandas as pd, matplotlib.pyplot as plt, seaborn as sns
sns.set_theme(style='whitegrid')
raw=pd.read_excel('../dataset/Tourism_Raw.xlsx')
print(raw.shape); print(raw.isnull().sum()); print('Duplicates:',raw.duplicated().sum())
```

## Data Cleaning

```python
df=raw.drop_duplicates().copy()
df=df[df.stay_nights>0]
df['spend_usd']=df.spend_usd.fillna(df.groupby('destination').spend_usd.transform('median'))
df['rating']=df.rating.fillna(df.rating.median())
df['travel_date']=pd.to_datetime(df.travel_date)
df['year']=df.travel_date.dt.year; df['month']=df.travel_date.dt.month
df.to_excel('../dataset/Tourism_Cleaned_Preprocessed.xlsx',index=False)
print('Cleaned:',df.shape); df.describe()
```

## 1. Top Destinations

```python
s=df.destination.value_counts().sort_values()
s.plot.barh(color='teal',figsize=(8,5),title='Top Destinations by Bookings')
plt.tight_layout(); plt.savefig('../visualizations/top_destinations.png',dpi=150); plt.show()
```

## 2. Monthly Travel Trend

```python
t=df.groupby(df.travel_date.dt.to_period('M')).size()
t.index=t.index.to_timestamp()
t.plot(figsize=(10,4),marker='o',title='Monthly Bookings Trend')
plt.tight_layout(); plt.savefig('../visualizations/monthly_trend.png',dpi=150); plt.show()
```

## 3. Seasonality

```python
df.groupby('month').size().plot.bar(color='orange',figsize=(8,4),title='Bookings by Month (Seasonality)')
plt.tight_layout(); plt.savefig('../visualizations/seasonality.png',dpi=150); plt.show()
```

## 4. Tourist Origin Countries

```python
df.origin_country.value_counts().plot.bar(color='steelblue',figsize=(8,4),title='Tourists by Origin Country')
plt.tight_layout(); plt.savefig('../visualizations/origin_countries.png',dpi=150); plt.show()
```

## 5. Average Spend by Destination

```python
df.groupby('destination').spend_usd.mean().sort_values().plot.barh(color='purple',figsize=(8,5),title='Average Spend (USD) per Trip')
plt.tight_layout(); plt.savefig('../visualizations/avg_spend.png',dpi=150); plt.show()
```

## 6. Travel Type & Purpose

```python
pd.crosstab(df.travel_type,df.purpose).plot.bar(stacked=True,figsize=(8,5),title='Travel Type vs Purpose')
plt.tight_layout(); plt.savefig('../visualizations/type_purpose.png',dpi=150); plt.show()
```

## 7. Correlation Heatmap

```python
sns.heatmap(df[['stay_nights','spend_usd','rating','month']].corr(),annot=True,cmap='coolwarm')
plt.title('Correlation Heatmap'); plt.tight_layout(); plt.savefig('../visualizations/correlation_heatmap.png',dpi=150); plt.show()
```

## Key Findings
- Peak season: Oct-Dec and Apr-Jun
- Goa and Kerala lead bookings
- International tourists spend more per trip
- Spend is strongly linked to length of stay
